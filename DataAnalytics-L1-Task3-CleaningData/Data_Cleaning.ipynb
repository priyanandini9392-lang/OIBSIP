```python
print("Missing values before cleaning:")
print(df.isnull().sum())
```

```python
age_missing_before = df["Age"].isnull().sum()

df["Age"] = df["Age"].fillna(df["Age"].median())

age_missing_after = df["Age"].isnull().sum()

print("Missing Age values before:", age_missing_before)
print("Missing Age values after:", age_missing_after)
```

```python
### Missing Value Handling – Age

The Age column contained 177 missing values. Median imputation was used because Age is a numerical variable and the median is less affected by extreme values than the mean. This allows us to retain all passenger records without deleting rows.
```

```python
embarked_missing_before = df["Embarked"].isnull().sum()

df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

embarked_missing_after = df["Embarked"].isnull().sum()

print("Missing Embarked values before:", embarked_missing_before)
print("Missing Embarked values after:", embarked_missing_after)
```

```python
### Missing Value Handling – Embarked

The Embarked column contained 2 missing values. Since Embarked is a categorical variable, the mode was used for imputation. The most frequently occurring embarkation port was used to fill the missing records.
```

```python
cabin_missing_before = df["Cabin"].isnull().sum()

df["Cabin"] = df["Cabin"].fillna("Unknown")

cabin_missing_after = df["Cabin"].isnull().sum()

print("Missing Cabin values before:", cabin_missing_before)
print("Missing Cabin values after:", cabin_missing_after)
```

```python
### Missing Value Handling – Cabin

The Cabin column contained 687 missing values. Instead of deleting a large number of passenger records, the missing values were replaced with the label "Unknown". This preserves the passenger records while clearly indicating that cabin information was unavailable.
```

```python
print("Missing values after missing-data handling:")
print(df.isnull().sum())
```

```python
duplicates_before = df.duplicated().sum()

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicates_before)
print("Duplicate rows removed:", duplicates_before - duplicates_after)
print("Duplicate rows after cleaning:", duplicates_after)
```

```python
df["Sex"] = df["Sex"].astype(str).str.strip().str.title()

df["Embarked"] = df["Embarked"].astype(str).str.strip().str.upper()

print("Sex values:")
print(df["Sex"].unique())

print("\nEmbarked values:")
print(df["Embarked"].unique())
```

```python
### Data Standardization

Text fields were standardized to maintain consistent formatting. The Sex column was converted to title case, while the Embarked column was converted to uppercase. Leading and trailing spaces were also removed.
```

```python
df["PassengerId"] = df["PassengerId"].astype(str)

df["Survived"] = pd.to_numeric(df["Survived"], errors="coerce").astype("Int64")

df["Pclass"] = pd.to_numeric(df["Pclass"], errors="coerce").astype("Int64")

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

df["SibSp"] = pd.to_numeric(df["SibSp"], errors="coerce").astype("Int64")

df["Parch"] = pd.to_numeric(df["Parch"], errors="coerce").astype("Int64")

df["Fare"] = pd.to_numeric(df["Fare"], errors="coerce")

print("Corrected Data Types:")
print(df.dtypes)
```

```python
def find_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]
    
    return outliers, lower_bound, upper_bound


outlier_results = {}

for column in ["Age", "SibSp", "Parch", "Fare"]:
    outliers, lower, upper = find_outliers_iqr(df, column)
    
    outlier_results[column] = len(outliers)
    
    print(column)
    print("Lower Bound:", round(lower, 2))
    print("Upper Bound:", round(upper, 2))
    print("Number of Outliers:", len(outliers))
    print()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df["Fare"].dropna())

plt.title("Fare Outlier Detection")
plt.ylabel("Fare")

plt.show()
```

```python
### Outlier Handling Decision

Outliers were detected using the IQR method. The detected outliers were retained because they may represent genuine passenger records rather than incorrect data. For example, higher ticket fares can be legitimate for certain passengers. Therefore, no records were removed solely because they were identified as outliers.
```
