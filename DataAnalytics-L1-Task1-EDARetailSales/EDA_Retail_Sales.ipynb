```python
##Monthly Sales Trend
```

```python
# Monthly Sales

df['Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = (
    df.groupby('Month')['Sales']
    .sum()
    .sort_index()
)

print("Monthly Sales:")
print(monthly_sales)
```

```python
##Monthly Sales Graph
```

```python
# Monthly Sales Trend

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o'
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
### Observation

Monthly sales show variations over time, with some months recording significantly higher sales than others. This indicates changes in customer demand and sales performance across different periods.
```

```python
##Quarterly Sales
```

```python
# Quarterly Sales

df['Quarter'] = df['Order Date'].dt.to_period('Q')

quarterly_sales = (
    df.groupby('Quarter')['Sales']
    .sum()
    .sort_index()
)

print("Quarterly Sales:")
print(quarterly_sales)
```

```python
##Quarterly Sales Graph
```

```python
# Quarterly Sales Trend

plt.figure(figsize=(12, 6))

plt.plot(
    quarterly_sales.index.astype(str),
    quarterly_sales.values,
    marker='o'
)

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
### Observation

Quarterly sales show variations in overall business performance across different quarters. Some quarters generate significantly higher sales than others, indicating changes in customer demand and sales performance over time.
```

```python
##Correlation Analysis
```

```python
# Correlation Analysis

correlation_columns = [
    'Sales',
    'Quantity',
    'Discount',
    'Profit',
    'Shipping Cost'
]

correlation = df[correlation_columns].corr()

print("Correlation Matrix:")
print(correlation)
```

```python
##Correlation Heatmap
```

```python
# Correlation Heatmap

plt.figure(figsize=(8, 6))

plt.imshow(
    correlation,
    cmap='coolwarm',
    aspect='auto'
)

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Heatmap")

for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(
            j,
            i,
            f"{correlation.iloc[i, j]:.2f}",
            ha='center',
            va='center'
        )

plt.tight_layout()
plt.show()
```

```python
### Observation

The correlation heatmap shows that Sales has a moderate positive relationship with Quantity and Profit. Sales also has a relatively strong positive relationship with Shipping Cost. Discount has a weak negative relationship with Sales and a moderate negative relationship with Profit. This indicates that higher discounts may be associated with lower profitability.
```
